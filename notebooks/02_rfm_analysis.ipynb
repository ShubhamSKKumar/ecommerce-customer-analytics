# RFM Customer Segmentation
Analyzing Recency, Frequency, and Monetary value of customers.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```
rfm = pd.read_csv('../data/cleaned/customer_segments.csv')
rfm.head()
```

## Segment Distribution

```
segment_counts = rfm['Segment'].value_counts()
plt.figure(figsize=(10, 6))
sns.barplot(y=segment_counts.index, x=segment_counts.values, palette='coolwarm')
plt.title('Customer Distribution by RFM Segment')
plt.xlabel('Number of Customers')
plt.show()
```

## RFM Value by Segment

```
rfm_agg = rfm.groupby('Segment').agg({
    'Recency': 'mean',
    'Frequency': 'mean',
    'Monetary': 'mean',
    'CustomerID': 'count'
}).round(1)
rfm_agg.rename(columns={'CustomerID': 'Count'}, inplace=True)
rfm_agg.sort_values('Monetary', ascending=False)
```
